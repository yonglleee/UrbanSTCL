```python
import pandas as pd

df = pd.read_pickle('/data_nas/huangyj/placePulse/pulse_score_path_42045.p')

# 剔除 safety 列 NaN 的值
df_cleaned = df.dropna(subset=['safety'])

# 统计不含 NaN 的数量
count_non_nan = df_cleaned['safety'].count()

print(count_non_nan)
```

```python
import matplotlib.pyplot as plt

# 保留两位小数
df_cleaned['safety'] = df_cleaned['safety'].round(2)

# 绘制直方图
plt.figure(figsize=(10, 6))
plt.hist(df_cleaned['safety'].dropna(), bins=20, edgecolor='black')
plt.xlabel('Safety Values')
plt.ylabel('Frequency')
plt.title('Histogram of Safety Values')
plt.grid(axis='y')
plt.show()
```

```python
# 将 'mode' 列重命名为 'split'
df_cleaned.rename(columns={ 'label':'safety'}, inplace=True)
# 导出为 .p 文件

```

```python
# 添加 'label' 列
df_cleaned['label'] = -1  # 默认值为 -1

# 根据 safety 列的值进行条件赋值
df_cleaned.loc[df_cleaned['safety'] < 3.5, 'label'] = 0
df_cleaned.loc[df_cleaned['safety'] > 6.5, 'label'] = 1
```

```python
# 筛选 label 大于 0 的行
filtered_df = df_cleaned[df_cleaned['label'] >= 0]

root = '/data_nas/huangyj/placePulse/rawData/googleviews-110802files/'

filtered_df['path'] = root + filtered_df['name'] 

# 导出为 .p 文件
filtered_df.to_pickle('filtered_labels.p')
```

```python
# 加载CLIP特征文件
file_path = '/home/liyong/code/svpretrain/Perception/feature/clip.pkl'
df = pd.read_pickle(file_path)
print(f"加载了{len(df)}个样本")

# 显示更新前的路径示例
print("\n更新前的路径示例:")
for i, path in enumerate(df['path'][:5]):
    print(f"{i+1}. {path}")

# 更新路径
original_path = '/data_nas/huangyj/placePulse/'
new_path = '/data_nas/huangyj/data/placePulse/'

# 替换路径
print("正在更新路径...")
df['path'] = df['path'].apply(lambda x: x.replace(original_path, new_path) if isinstance(x, str) else x)

# 显示更新后的路径示例
print("\n更新后的路径示例:")
for i, path in enumerate(df['path'][:5]):
    print(f"{i+1}. {path}")

# 保存更新后的文件
output_path = '/home/liyong/code/svpretrain/Perception/feature/clip_updated.pkl'
df.to_pickle(output_path)
print(f"\n已保存更新后的文件到: {output_path}")

# 检查更新后的路径是否存在
import os
from tqdm import tqdm

# 检查所有路径
exists_count = sum(1 for path in tqdm(df['path'], desc="检查文件") if os.path.exists(path))
missing_count = len(df) - exists_count

print(f"\n路径检查结果:")
print(f"总路径数: {len(df)}")
print(f"存在的文件: {exists_count} ({exists_count/len(df)*100:.2f}%)")
print(f"缺失的文件: {missing_count} ({missing_count/len(df)*100:.2f}%)")
```
